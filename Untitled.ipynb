```python
import os
import matplotlib.pyplot as plt
from linear_classifier_3classes import MLP
import torch

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
pixel_size = 28
clf_path = "/vol/biomedic3/rrr2417/emergent_trajectories/models/classifier/shapes/model.pt"
INPUT_DIM = pixel_size * pixel_size * 3

classifier = MLP(INPUT_DIM, [2, 2, 2])
classifier.load_state_dict(torch.load(clf_path, weights_only=True))
classifier = classifier.to(device)
```

```python
from itertools import product
import re
import numpy as np
from torchvision.utils import make_grid
import torchvision.transforms.functional as F

files = os.listdir(folder)

accuracies = {}

for image_cond in product((0, 1), (0, 1), (0, 1)):
    print(image_cond)
    accuracies[image_cond] = []
    image_files = [
        (os.path.join(folder, x), int(re.search(r"ep(\d+)", x).group(1)))
        for x in files if str(image_cond) in x
    ]
    image_files = sorted(image_files, key=lambda x: x[1])
    x_gens = [
        (torch.from_numpy(np.clip(np.load(x)["x_gen"], 0, 1)).to(device), epoch)
        for x, epoch in image_files
    ]
    # print(len(x_gens))
    # print(x_gens[60][0][:16].shape)
    # grid = make_grid(x_gens[80][0][:16], nrow=4)
    # plt.imshow(F.to_pil_image(grid))
    # plt.show()
    with torch.no_grad():
        for x_gen, epoch in x_gens:
            y_pred = classifier(x_gen)
            accs = []
            for ii in range(3):
                top_pred = y_pred[ii].cpu().numpy().argmax(1)
                acc = np.array(top_pred == int(image_cond[ii]), dtype=np.int32)
                accs.append(acc)
            acc = np.mean(accs)
            accuracies[image_cond].append((epoch, acc))
```

```python
plt.xscale("log")

label_keys = {
    (0, 0, 0): {"color": "tab:cyan", "linestyle": "solid"},
    (0, 0, 1): {"color": "tab:cyan", "linestyle": "dashed"},
    (0, 1, 0): {"color": "tab:cyan", "linestyle": "dashdot"},
    (1, 0, 0): {"color": "tab:cyan", "linestyle": "dotted"},
    (0, 1, 1): {"color": "tab:pink", "linestyle": "solid"},
    (1, 0, 1): {"color": "tab:pink", "linestyle": "dashed"},
    (1, 1, 0): {"color": "tab:pink", "linestyle": "dashdot"},
    (1, 1, 1): {"color": "tab:green", "linestyle": "solid"},
}

for k, plot_attr in label_keys.items():
    v = accuracies[k]
    xs = [x for x, _ in v][::5]
    ys = [y for _, y in v][::5]
    plt.plot(xs, ys, color=plot_attr["color"], linestyle=plot_attr["linestyle"], label=k)
plt.legend()
plt.show()
```
